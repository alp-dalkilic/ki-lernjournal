```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression

X = np.array([[0.5, 1.5], [1, 1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1])
X_neu = np.array([[2.5, 1], [0.5, 0.5]])
# Klasse LogistischeRegression aus Zelle [0] muss ausgeführt sein
class LogistischeRegression:
    def __init__(self, alpha=0.1, n_iter=1000):
        # Hyperparameter speichern; self.w = None, self.b = None, self.J_verlauf = []
        self.alpha = alpha
        self.n_iter = n_iter
        self.w = None
        self.b = None
        self.J_verlauf = []

    def sigmoid(self, z):
        # 1 / (1 + np.exp(-z))
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):
        m, n = X.shape
        w = np.zeros(n) 
        b = 0.0
        for i in range(self.n_iter):
            p = self.sigmoid(X @ w + b)
            err = p - y
            w -= self.alpha * (X.T @ err) / m
            b -= self.alpha * err.mean()
            J = -np.mean(y*np.log(p) + (1-y)*np.log(1-p))
            self.J_verlauf.append(J)
        self.w = w
        self.b = b

        return self

    def predict_proba(self, X):
        return self.sigmoid(X @ self.w + self.b)

    def predict(self, X):
        return (self.predict_proba(X) >= 0.5).astype(int)

eigen = LogistischeRegression(alpha=0.1, n_iter=1000).fit(X, y)
sk = LogisticRegression(max_iter=1000).fit(X, y)
print(eigen.w.shape, eigen.b, sk.coef_.shape, sk.intercept_.shape, sk.n_iter_)
vergleich = pd.DataFrame({'eigen': eigen.w, 'sklearn': sk.coef_[0]}, index=['x1', 'x2'])
print(vergleich)
gl_vorz = (np.sign(vergleich['eigen']) == np.sign(vergleich['sklearn'])).sum() 
print(gl_vorz)
```
