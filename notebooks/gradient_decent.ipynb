```python
x = [1, 2, 3]
y = [3, 5, 7]
w_start = 0
b_start = 0
iteration = 10
alpha = 0.1

def total_cost(x, y, w, b):
    sum_sq_cost = 0
    m = len(x)
    for i in range(m):
        sq_cost = (w*x[i] + b - y[i])**2
        sum_sq_cost += sq_cost
    cost = sum_sq_cost / (2*m)

    return cost

def gradient(x, y, w, b):
    sum_dj_dw = 0
    sum_dj_db = 0
    m = len(x)
    for i in range(m):
        err = (w * x[i] + b - y[i])
        dj_dw_i = err * x[i]
        dj_db_i = err
        sum_dj_dw += dj_dw_i
        sum_dj_db += dj_db_i
    
    dj_dw = sum_dj_dw / m
    dj_db = sum_dj_db / m

    return dj_dw, dj_db

def gradient_descent(x, y, w_start, b_start, alpha, iteration):
    w = w_start
    b = b_start
    for i in range(iteration):
        gradient_w, gradient_b = gradient(x, y, w, b)
        w = w - alpha * gradient_w
        b = b - alpha * gradient_b

    return w, b

w, b = gradient_descent(x, y, w_start, b_start, 0.1, 10)
print(w, b, total_cost(x, y, w, b))
```
