```python
class Konto:
    def __init__(self, inhaber, stand):
        self.inhaber = inhaber
        self.stand = stand
        self.verlauf = []

    def einzahlen(self, betrag):
        if betrag <= 0:
            raise ValueError("Betrag muss > 0 sein")
        self.stand += betrag
        self.verlauf.append(betrag)

    def abheben(self, betrag):
        if betrag <= 0:
            raise ValueError("Betrag muss > 0 sein")
        if betrag > self.stand:
            raise ValueError(f'Kontostand: {self.stand} EUR nicht ausreichend')
        self.stand -= betrag
        self.verlauf.append(betrag*(-1))

    def saldo_text(self):
        return f'{self.inhaber}: {self.stand:.2f} EUR'

    def anzahl_buchungen(self):
        return len(self.verlauf)

k = Konto("Alp", 100)
k.einzahlen(50)
k.abheben(30)
assert k.stand == 120
assert k.anzahl_buchungen() == 2
assert k.saldo_text() == "Alp: 120.00 EUR"
try:
    k.abheben(500)
    assert False, "hätte ValueError werfen müssen"
except ValueError:
    pass
assert k.stand == 120          # Fehlversuch hat nichts verändert
print("Konto ok")
```
