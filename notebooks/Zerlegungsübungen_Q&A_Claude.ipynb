```python
# Gleiche Aufgabe, wie die zuvor nur mit einer anderen Logik ermittelt
kunden        = ["Anna", "Ben", "Cem", "Anna", "Cem", "Ben", "Anna"]
bestellungen  = [45, 120, 30, 200, 95, 150, 60]

# Teil 1: Summen-Dict bauen (identisch zu deinem)
summen = {}
for i in range(len(kunden)):
    if kunden[i] not in summen:
        summen[kunden[i]] = bestellungen[i]   # neuer Kunde -> anlegen
    else:
        summen[kunden[i]] += bestellungen[i]  # bekannter Kunde -> dazuaddieren

# Teil 2: argmax - "besten bisher" im Gleichklang mitfuehren
top_kunde = ""     # Platzhalter fuer den Namen
top_summe = 0      # Startwert (hier ok, weil Summen immer positiv sind)

for kunde in summen:                 # laeuft ueber die Schluessel (Namen)
    if summen[kunde] > top_summe:    # schlaegt dieser Kunde die Bestmarke?
        top_summe = summen[kunde]    # dann BEIDE zusammen aktualisieren
        top_kunde = kunde

print(top_kunde, top_summe)          # -> Anna 305
```
