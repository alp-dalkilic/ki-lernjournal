```python
# Trinkgeldrechner (nach Claude Feedback)
def pro_person(rechnungsbetrag, tip, anzahl_pers):
    gesamt = rechnungsbetrag * (1+tip/100)
    gesamt_person = round(gesamt / anzahl_pers, 2)
    return gesamt, gesamt_person

while True:
    try:
        
        rechnungsbetrag = float(input("Rechnungsbetrag exkl. Trinkgeld: "))
        
        tip = float(input("Trinkgeld in Prozent, e.g. nur Zahl 7.5 für 7.5%' eingeben: "))
        
        anzahl_pers = int(input("Anzahl Personen: "))

        if rechnungsbetrag < 0 or tip < 0 or anzahl_pers <= 0:
            print("Ungültige Eingabe! Keine negativen Zahlen - Min 1 Person als Ganzzahl")
            continue

        break

    except ValueError:
        print("Ungültige Eingabe! Nur Zahlen ohne Text oder Zeichen - Punkt als Separator bei Nachkommastellen")

gesamt, gesamt_person = pro_person(rechnungsbetrag, tip, anzahl_pers)
print(round(gesamt, 2), gesamt_person)
```

```python
# Trinkgeldrechner (erste Versuch)
while True:
    try:
        
        rechnungsbetrag = float(input("Rechnungsbetrag exkl. Trinkgelt: "))
        
        tip = int(input("Trinktgeld in Prozent, e.g. nur Ganzzahl 5 für 5%' eingeben: "))
        
        anzahl_pers = int(input("Anzahl Personen: "))
    
        gesamt = round(rechnungsbetrag * (1+tip/100), 2)
        gesamt_person = round(gesamt / anzahl_pers, 2)

        break

    except (ValueError, ZeroDivisionError):
        print("Ungültige Einfabe! Min 1 Person - Nur Zahlen ohne Text oder Zeichen - Punkt als Separator bei Nachkommastellen")

print(gesamt, gesamt_person)
```
