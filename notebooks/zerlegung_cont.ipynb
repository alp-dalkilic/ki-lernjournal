```python
kunden = ["Anna", "Ben", "Cem", "Anna", "Cem", "Ben", "Anna"]
bestellungen = [45, 120, 30, 200, 95, 150, 60]

kd_uebersicht = {}

for i in range(len(kunden)):
    kunde = kunden[i]
    bestellung = bestellungen[i]

    if kunde not in kd_uebersicht:
        kd_uebersicht[kunde] = [bestellung]
    else:
        kd_uebersicht[kunde].append(bestellung)

print(kd_uebersicht)
```

```python
produkte    = ["Apfel", "Brot", "Banane", "Käse", "Birne", "Milch"]
kategorien  = ["Obst", "Backwaren", "Obst", "Molkerei", "Obst", "Molkerei"]

p_gruppierungen = {}

for i in range(len(kategorien)):

    kat = kategorien[i]
    prod = produkte[i]

    if kat not in p_gruppierungen:
        p_gruppierungen[kat] = [prod]

    else:
        p_gruppierungen[kat].append(prod)

print(p_gruppierungen)
```

```python
bestellungen = [
    ("Meier",  45),
    ("Huber", 120),
    ("Meier",  30),
    ("Gruber", 200),
    ("Huber",  95),
    ("Meier", 150),
]

kd_logbuch = {}

for name, betrag in bestellungen:

    if name not in kd_logbuch:
        kd_logbuch[name] = 0
    kd_logbuch[name] += betrag

beste_best = max(kd_logbuch, key=kd_logbuch.get)

print(beste_best)
print(kd_logbuch[beste_best])


```

```python
bestellungen = [
    ("Meier",  45),
    ("Huber", 120),
    ("Meier",  30),
    ("Gruber", 200),
    ("Huber",  95),
    ("Meier", 150),
]

kd_logbuch = {}

for name, betrag in bestellungen:

    if name not in kd_logbuch:
        kd_logbuch[name] = 0
    kd_logbuch[name] += betrag


b_name = ""
b_umsatz = 0

for name, umsatz in kd_logbuch.items():
    if b_name == "" or b_umsatz < umsatz:
        b_name = name
        b_umsatz = umsatz

print(b_name, b_umsatz)
```
