```python
# Task 5: Namens-Normalisierer

ergebnis = []
namensliste = ["   mueller, anna ", "SCHMIDT  , ben", "mueller,anna", ""]
for eintrag in namensliste:
    items = eintrag.split(",")

    if len(items) == 2:
        vorname = items[1].strip().title()
        nachname = items[0].strip().title()

        name = " ".join([vorname, nachname])
        if name not in ergebnis:
            ergebnis.append(name)

ergebnis.sort()

print(ergebnis)

```

```python
# Task 5: Namens-Normalisierer (Alternativversion ohne join)

ergebnis = []
namensliste = ["   mueller, anna ", "SCHMIDT  , ben", "mueller,anna", ""]
for eintrag in namensliste:
    items = eintrag.split(",")

    if len(items) == 2:
        vorname = items[1].strip().title()
        nachname = items[0].strip().title()

        name = f"{vorname} {nachname}"
        if name not in ergebnis:
            ergebnis.append(name)

ergebnis.sort()

print(ergebnis)

```
