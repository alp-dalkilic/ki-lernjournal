```python
# Wort-Haeufigkeitszaehler

text = "Die Bank prueft die Bank. Die Daten der Bank."
text_lc = text.lower()
text_clean = text_lc.replace(".", "")
liste = text_clean.split()

haeufigkeit = {}
for word in liste:
    if word not in haeufigkeit:
        haeufigkeit[word] = 1
    else:
        haeufigkeit[word] += 1

uniquewortzahl = len(haeufigkeit)

gesamtwortzahl = sum(haeufigkeit.values())

def sortwert(eintrag):
    return eintrag[1]

sortiert_dict = dict(sorted(haeufigkeit.items(), key=sortwert, reverse=True))


sortiert_list = list(sortiert_dict.items())

def top_fuenf(sortiert_list):
    if len(sortiert_list) <= 5:
        return sortiert_list

    grenzwert = sortiert_list[4][1]

    ergebnis = []

    for key, value in sortiert_list:# Hier ist der Code bewusst so geschrieben, dass mehrere mit ins Ranking kommen, wenn sie den gleichen Wert haben wie Pos 5 der Top 5
        if value >= grenzwert:
            ergebnis.append((key, value))
    return ergebnis

ergebnis = top_fuenf(sortiert_list)

print(f"Der bereinigte Text als Liste:\n{liste}")
print(f"Die Zähl-Dictionary:\n{haeufigkeit}")
print(f"Gesamtwortzahl: {gesamtwortzahl}, Uniquewortzahl: {uniquewortzahl}")
print(f"Die sortierte Dictionary:\n{sortiert_dict}")
print(f"Die sortierte Liste: \n{sortiert_list}")
print(f"Die Top-5 Wörter sind: \n{ergebnis}")

```
