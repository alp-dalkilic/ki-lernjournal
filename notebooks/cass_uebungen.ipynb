```python
class Sparschwein:
    def __init__(self, name):
        self.name = name
        self.inhalt = 0
        self.einwuerfe = 0

    def einwerfen(self,betrag):
        self.inhalt += betrag
        self.einwuerfe += 1

    def stand(self):
        return self.name + ":" + str(self.inhalt) + " Euro nach " + str(self.einwuerfe) + " Einwürfen"

s = Sparschwein("Urlaubskasse")
s.einwerfen(20)
s.einwerfen(5)
print(s.stand())      # Urlaubskasse: 25 Euro nach 2 Einwürfen
print(s.inhalt)
```

```python
import numpy as np
class Zentrierer:
    def __init__(self):
        self.mittel = None

    def fit(self, X):
        self.mittel = np.mean(X, axis=0)

    def transform(self, X):
        return X - self.mittel
    
X = np.array([[1.0, 10.0],
              [3.0, 20.0],
              [5.0, 30.0]])

X_neu = np.array([[4.0, 15.0],
                  [6.0, 25.0]])

z = Zentrierer()
z.fit(X)
print(z.mittel)            # erwartet: [ 3. 20.]
print(z.transform(X))      # erwartet: [[-2. -10.] [ 0.  0.] [ 2. 10.]]
print(z.transform(X).mean(axis=0))   # erwartet: [0. 0.]

print(z.transform(X_neu))               # erwartet: [[-1. -5.] [ 1.  5.]]
print(z.transform(X_neu).mean(axis=0))  # erwartet: [0. 0.]?  — nein. Welcher Wert kommt?
```
