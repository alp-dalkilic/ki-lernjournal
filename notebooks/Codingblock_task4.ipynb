```python
# Codeblock Task 5: adress book

adressbuch = {}

def hinzufuegen(adressbuch):
    while True:

        while True:
            name = input(f"Geben sie ihren Vor- und Nachnamen ein: ")
            if name in adressbuch.keys():
                print(f"Es gibt diesen Namen bereits im Adressbuch")
            else:
                break
        
        adressbuch[name] = {"mail": input(f"Geben sie bitte ihre Email-Adresse ein: "),
                            "telefon": input(f"Geben sie bitte ihre Telefonnummer mit Ländervorwahl ein, z.B. +49 für Deutschland: "),
                            "ort": input(f"Geben sie bitte ihren Wohnort ein: ")}

        weiter = input(f"Möchten sie noch einen Kontakt hinzufügen (j/n): ")
        if weiter.lower() != "j":
            break

def suchen(adressbuch):
    name = input(f"Geben sie den Namen des gesuchten Kontakts ein: ")
    kontakt = adressbuch.get(name)
    
    if kontakt:
        print(name)
        print(kontakt)
    else:
        print("Kontakt nicht im Adressbuch vorhanden")

def auflisten(adressbuch):
    for k in sorted(adressbuch):
        print(k, adressbuch[k])

def loeschen(adressbuch):
    l_person = input(f"Geben sie den Namen des zu löschenden Kontakts ein: ")
    kontakt = adressbuch.get(l_person)

    if kontakt:
        del adressbuch[l_person]
        print(f"Kontakt: '{l_person}' wurde gelöscht")
    else:
        print(f"Kontakt nicht im Adressbuch vorhanden")

print("ADRESSBUCH")
print("Es stehen 5 Optionen zur Verfügung:")
print("1 Kontakt hinzufuegen")
print("2 Kontakt suchen")
print("3 Kontakte auflisten")
print("4 Kontakt löschen")
print("5 Adressbuch schließen")
print("Geben sie die jeweile Zahl für die gewünschte Option ein")

while True:
    try:
        option = int(input("Geben sie die gewünschte Option (Ganzzahl 1 - 5) ein: "))
        if option not in range(1, 6):
            print("Eingabe muss eine Zahl zwischen 1 und 5 sein")
        elif option == 5:
            break
        elif option == 1:
            hinzufuegen(adressbuch)
        elif option == 2:
            suchen(adressbuch)
        elif option == 3:
            auflisten(adressbuch)
        elif option == 4:
            loeschen(adressbuch)
    except ValueError:
        print("Eingabe muss eine Zahl zwischen 1 und 5 sein")

```
