```python
# Codeblock Task 5: adress book

adressbuch = {}

def hinzufuegen(adressbuch):
    while True:

        while True:
            name = input(f"Geben sie ihren Vor- und Nachnamen ein: ")
            if name in adressbuch.keys():
                print(f"Es gibt diesen Namen bereits im Adressbuch")
            else:
                break
        
        adressbuch[name] = {"mail": input(f"Geben sie bitte ihre Email-Adresse ein: "),
                            "telefon": input(f"Geben sie bitte ihre Telefonnummer mit Ländervorwahl ein, z.B. +49 für Deutschland: "),
                            "ort": input(f"Geben sie bitte ihren Wohnort ein: ")}

        weiter = input(f"Möchten sie noch einen Kontakt hinzufügen (j/n): ")
        if weiter.lower() != "j":
            break

def suchen(adressbuch):
    name = input(f"Geben sie den Namen des gesuchten Kontakts ein: ")
    kontakt = adressbuch.get(name)
    
    if kontakt:
        print(name)
        print(kontakt)
    else:
        print("Kontakt nicht im Adressbuch vorhanden")


hinzufuegen(adressbuch)
suchen(adressbuch)
```
