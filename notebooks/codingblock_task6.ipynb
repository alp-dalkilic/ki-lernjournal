```python
# Task 6: Parser als Funktion formuliert

def parse_ergebnis(block):
    ergebnis = []
    for eintrag in block.split("\n"):
        if eintrag == "":
            continue
        teile = eintrag.split(";")
       
        datum_w = teile[0]
        kategorie_w = teile[1]
        betrag_w = float(teile[2])

        dict_items = {"datum": datum_w, "kategorie": kategorie_w, "betrag": betrag_w}
        ergebnis.append(dict_items)
    return ergebnis

block = """01.03.2026;Lebensmittel;42.50
03.03.2026;Transport;19.90
05.03.2026;Lebensmittel;7.20
12.04.2026;Miete;850.00
14.04.2026;Transport;19.90
"""

for eintrag in parse_ergebnis(block):
    print(eintrag)
```
