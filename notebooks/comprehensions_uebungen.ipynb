```python
block = """01.03.2026;Lebensmittel;42.50
03.03.2026;Transport;19.90
05.03.2026;Lebensmittel;7.20
12.04.2026;Miete;850.00
14.04.2026;Transport;19.90
"""

belege = []
for line in block.splitlines():
    if line.strip() == '':
        continue
    d, k, b = line.split(';')
    belege.append({'datum': d, 'kategorie': k, 'betrag': float(b)})

daten  = [[2, 8], [8, 9], [8, 13]]
mittel = [6.0, 10.0]
X      = [[zeile[j] - mittel[j] for j in range(2)] for zeile in daten]
n      = len(X)

print(belege[0])
print(X)
```

```python
umgepackt = [[beleg['betrag'], beleg['datum'], beleg['kategorie']] for beleg in belege]
umgepackt
```

```python
ueber_20 = [beleg['betrag'] for beleg in belege if beleg['betrag'] > 20]
ueber_20
```

```python
ersetzt = [beleg['betrag'] if beleg['betrag'] >= 10 else 0 for beleg in belege]
ersetzt
```

```python
gefiltert = ['gross' if beleg['betrag'] > 100 else beleg['betrag'] for beleg in belege if beleg['betrag'] > 20]
gefiltert
```

```python
summe = sum([X[j][0]**2 for j in range(n)])
summe
```
