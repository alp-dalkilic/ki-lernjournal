```python
import numpy as np

# ---------- Datensatz A: Ngs Testdaten ----------
X_ng = np.array([[0.5, 1.5],
                 [1.0, 1.0],
                 [1.5, 0.5],
                 [3.0, 0.5],
                 [2.0, 2.0],
                 [1.0, 2.5]])
y_ng = np.array([0, 0, 0, 1, 1, 1])

# ---------- Datensatz B: Ausfall-Scoring (aus Uebung_ausfalls_scoring.ipynb) ----------
# Spalte 0: Einkommen (Tsd. €), Spalte 1: Verschuldungsquote (%)
X_ausfall = np.array([[62, 25],
                      [38, 48],
                      [55, 30],
                      [29, 55],
                      [71, 20],
                      [44, 42],
                      [33, 50],
                      [58, 35]], dtype=float)
y_ausfall = np.array([0, 1, 0, 1, 0, 1, 1, 0])   # 1 = Ausfall

w = np.array([2., 3.])
b = 1.
X, y = X_ng, y_ng

# ---------- deine Funktionen vom 02.10. (unverändert übernommen) ----------
def standardisiere(X):
    mu = np.mean(X, axis=0)
    sigma = np.std(X, axis=0)
    X_std = (X - mu) / sigma
    return mu, sigma, X_std

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

mu, sigma, X_std = standardisiere(X_ng)
```

```python
def gradient_logistic(X_std, y, w, b):
    z = X @ w + b
    p = sigmoid(z)
    m = len(y)
    err = p - y
    dj_dw = err @ X / m
    dj_db = err.mean()
    return dj_dw, dj_db

dj_dw, dj_db = gradient_logistic(X_std, y, w, b)
print(dj_dw)
print(dj_db)
```
