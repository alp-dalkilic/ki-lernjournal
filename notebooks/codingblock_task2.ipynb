```python
# Zahlenfolge-Analyse - nach Feedback

from pprint import pprint

liste = [3,1,4,1,5,9,2,6]

def rechnen(liste):
    summe = sum(liste)
    mittelwert = round(summe/len(liste),2)
    minimum = min(liste)
    maximum = max(liste)    
    return summe, mittelwert, minimum, maximum


def zaehlen(liste):
    gerade = 0
    ungerade = 0
    for zahlen in liste:
        if zahlen % 2 == 0:
            gerade += 1
        else:
            ungerade += 1
    return gerade, ungerade

def serie(liste):
    laenge = 1
    max_laenge = 1
    for i in range(1,len(liste)):
        if liste[i] > liste[i-1]:
            laenge += 1
            if laenge > max_laenge:
                max_laenge = laenge
        else:
            laenge = 1
    return max_laenge

def analysiere(liste):
    summe, mittelwert, minimum, maximum = rechnen(liste)
    gerade, ungerade = zaehlen(liste)
    max_serie = serie(liste)
    ergebnis = {"summe": summe, "mittelwert": mittelwert, "minimum": minimum, "maximum": maximum, "gerade": gerade, "ungerade": ungerade, "max_serie": max_serie}
    return ergebnis

try:
    ergebnis = analysiere(liste)
    pprint(ergebnis)
except ZeroDivisionError:
    print("Division mit Nenner gleich NULL wegen leerer List")
```

```python
# Zahlenfolge-Analyse - vor Feedback

from pprint import pprint

liste = [3,1,4,1,5,9,2,6]

def rechnen(liste):
    summe = sum(liste)
    mittelwert = summe/len(liste)
    minimum = min(liste)
    maximum = max(liste)    
    return summe, mittelwert, minimum, maximum

def zaehlen(liste):
    liste_gerade = []
    liste_ungerade = []
    for zahlen in liste:
        if zahlen % 2 == 0:
            liste_gerade.append(1)
        else:
            liste_ungerade.append(1)
        anzahl_ger = len(liste_gerade)
        anzahl_unger = len(liste_ungerade)
    return anzahl_ger, anzahl_unger

def serie(liste):
    serie = [1]
    max_serie = []
    for i in range(1,len(liste)):
        if liste[i] > liste[i-1]:
            serie.append(1)
            if len(serie) > len(max_serie):
                max_serie = serie
            else:
                serie = [0]
    return len(max_serie)

summe, mittelwert, minimum, maximum = rechnen(liste)
anzahl_ger, anzahl_unger = zaehlen(liste)
max_serie = serie(liste)

dict = {"Summe": summe, "Mittelwert": mittelwert, "Maximum": maximum, "Minimum": minimum, "No. Gerade Zahlen": anzahl_ger, "No. Ungerader Zahlen": anzahl_unger, "Längste Serie": max_serie}

pprint(dict)
```
