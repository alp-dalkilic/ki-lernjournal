```python
import numpy as np
X = np.array([[0.5, 1.5], [1, 1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1])
w = np.array([1.0, 1.0])
b = -3.0

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def kosten_logistisch(X, y, w, b):
    z = X @ w + b
    p = sigmoid(z)
    loss = -y * np.log(p) - (1-y) * np.log(1-p)
    cost = np.mean(loss)
    return cost
    
def vorhersage(X, w, b):
    z = X @ w + b
    prediction = (z >= 0).astype(int)
    #prediction = [1 if zi >= 0 else 0 for zi in z] ist eine Liste, daher die andere Form, die einen Array liefert und auch kürzer und einfacher ist.
    return prediction


print(kosten_logistisch(X, y, w, b))
print(vorhersage(X, w, b))
```
