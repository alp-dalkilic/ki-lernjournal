```python
import numpy as np
np.set_printoptions(precision=4, suppress=True)

D  = np.array([[2, 8], [8, 9], [8, 13]], dtype=float)
n  = D.shape[0]                       # 3 Zeilen
X  = D - D.mean(axis=0)               # zentriert       (Schritt 3)
Xs = X / D.std(axis=0, ddof=1)        # standardisiert  (Schritt 4)
C  = X.T @ X / (n-1)                  # Kovarianzmatrix (Schritt 5)

print(Xs)
```

```python
lams, vecs = np.linalg.eigh(C)
lams    # [ 3. 16.]
vecs    # [[ 0.5547 -0.8321]
        #  [-0.8321 -0.5547]]
```

```python
print(C @ vecs[:, 1])     # [-13.3128  -8.8752]
print(16 * vecs[:, 1])    # [-13.3128  -8.8752]
```

```python
S = Xs.T @ Xs / (n-1)
S
```

```python
lams, vecs = np.linalg.eigh(S)
lams_switch = lams[::-1]
vecs_switch = vecs[:, ::-1]
print(lams_switch)
print(vecs_switch)
```

```python
scores = Xs @ vecs_switch
print(scores)
```

```python
anteile = lams_switch * 100 / lams_switch.sum()
print(anteile)
```

```python
import numpy as np

A = np.array([[1., 1.],
              [1., 2.],
              [1., 3.]])
y = np.array([1., 2., 2.])

Q, R = np.linalg.qr(A)
print("R =\n", R.round(3))            # obere Dreiecksmatrix
print("QᵀQ =\n", (Q.T @ Q).round(10)) # Einheitsmatrix → Qᵀ ersetzt die Inverse

z = Q.T @ y                           # rechte Seite von R·w = Qᵀy

# Rückwärtseinsetzen: unterste Zeile zuerst
w = z[1] / R[1, 1]
b = (z[0] - R[0, 1] * w) / R[0, 0]
print("b =", round(b, 4), " w =", round(w, 4))
```

```python
print(np.linalg.cond(A))          # 6.79
print(np.linalg.cond(A.T @ A))    # 46.14  = 6.79²
```
