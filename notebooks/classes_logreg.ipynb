```python
import numpy as np

X = np.array([[0.5, 1.5], [1, 1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1])
X_neu = np.array([[2.5, 1], [0.5, 0.5]])

class LogistischeRegression:
    def __init__(self, alpha=0.1, n_iter=1000):
        # Hyperparameter speichern; self.w = None, self.b = None, self.J_verlauf = []
        self.alpha = alpha
        self.n_iter = n_iter
        self.w = None
        self.b = None
        self.J_verlauf = []

    def sigmoid(self, z):
        # 1 / (1 + np.exp(-z))
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):
        m, n = X.shape
        w = np.zeros(n) 
        b = 0.0
        for i in range(self.n_iter):
            p = self.sigmoid(X @ w + b)
            err = p - y
            w -= self.alpha * (X.T @ err) / m
            b -= self.alpha * err.mean()
            J = -np.mean(y*np.log(p) + (1-y)*np.log(1-p))
            self.J_verlauf.append(J)
        self.w = w
        self.b = b

        return self

    def predict_proba(self, X):
        return self.sigmoid(X @ self.w + self.b)

    def predict(self, X):
        return (self.predict_proba(X) >= 0.5).astype(int)


modell = LogistischeRegression(alpha=0.1, n_iter=1000).fit(X, y)
print(modell.w.round(4), modell.b.round(4), round(modell.J_verlauf[-1], 3))
# erwartet: [2.3046 2.0762] -5.9077 0.159

m2 = LogistischeRegression(alpha=0.1, n_iter=10000).fit(X, y)
print(m2.w.round(4), round(m2.b, 4))
# erwartet: [5.2812 5.0782] -14.2224

print(modell.predict(X_neu))
# erwartet: [1 0]
```

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

X_train = np.array([[0.5, 1.5], [1, 1], [1.5, 0.5],
                    [3, 0.5],   [2, 2], [1, 2.5]])
y_train = np.array([0, 0, 0, 1, 1, 1])

modell = LogisticRegression()          # Objekt anlegen = __init__
modell.fit(X_train, y_train)           # lernen = dein fit, nur mit y dazu

print(modell.coef_)                    # erwartet: [[0.904 0.736]]   ← w
print(modell.intercept_)               # erwartet: [-2.334]          ← b
print(modell.predict(X_train))         # erwartet: [0 0 0 1 1 1]
print(modell.score(X_train, y_train))  # erwartet: 1.0  (Anteil richtig)

# Was predict intern tut — dein z und dein Sigmoid von Tag 51:
z = X_train @ modell.coef_[0] + modell.intercept_[0]
print(1 / (1 + np.exp(-z)))            # erwartet: Werte < 0.5 für die ersten drei, > 0.5 für die letzten drei

# Regularisierung: C = 1/λ. Großes C = fast keine Regularisierung → große Gewichte
modell_frei = LogisticRegression(C=1e6)
modell_frei.fit(X_train, y_train)
print(modell_frei.coef_)               # erwartet: [[12.67 12.63]]
print(modell_frei.intercept_)          # erwartet: [-34.2]
```
