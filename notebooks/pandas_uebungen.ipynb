```python
import pandas as pd
import numpy as np

daten = {
    "kunde":     ["K01","K02","K03","K04","K05","K06","K07","K08","K09","K10"],
    "alter":     [34, 52, 23, 45, 61, 29, 38, 47, 26, 55],
    "einkommen": [3200, 5100, 1800, 4200, 2900, 2400, 3900, np.nan, 2100, 6000],
    "kredit":    [8000, 15000, 5000, 12000, 7000, 9000, 10000, 20000, 4000, 25000],
    "laufzeit":  [24, 36, 12, 48, 24, 36, 24, 60, 12, 48],
    "zweck":     ["Auto","Haus","Konsum","Haus","Auto","Konsum","Auto","Haus","Konsum","Haus"],
    "ausfall":   [0, 0, 1, 0, 0, 1, 0, 1, 1, 0],
}
df = pd.DataFrame(daten)
df
```

```python
#A1: Überblick - Anzahl Spalten, Daten, Typ je Spalte
df.info()
```

```python
#A2:Auswählen & Filtern
df.loc[df['ausfall'] == 1, ['kunde', 'einkommen', 'kredit']]
```

```python
#A3: Kennzahlen
df.ausfall.mean()
```

```python
#A4: Gruppieren
df.groupby('zweck').ausfall.mean()
```

```python
#A5: Fehlende Werte
display(df.einkommen.isnull().sum())
df['einkommen'] = df.einkommen.fillna(df.einkommen.mean())
display(df)
```

```python
#A6: Neue Spalte + Sortieren
df['rate'] = df.kredit / df.laufzeit
df.sort_values('rate', ascending=False).head(3)
```

```python
#A7: Umbenennen
renamed = df.rename(columns={'ausfall': 'ziel'})
display(renamed)
display(df)
```
