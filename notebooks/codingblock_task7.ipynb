```python
# Task 7: Transaktions-Auswerter(Capstone)

def parse_ergebnis(block):
    daten_tab = []
    for eintrag in block.split("\n"):
        if eintrag == "":
            continue
        teile = eintrag.split(";")
       
        datum_w = teile[0]
        kategorie_w = teile[1]
        betrag_w = float(teile[2])

        dict_items = {"datum": datum_w, "kategorie": kategorie_w, "betrag": betrag_w}
        daten_tab.append(dict_items)
    return daten_tab

def sum_cat(daten_tab):
    summe = {}
    for eintrag in daten_tab:
        kategorie = eintrag["kategorie"]
        betrag = eintrag["betrag"]

        summe[kategorie] = summe.get(kategorie, 0) + betrag

    for kategoie in summe:
        summe[kategorie] = round(summe[kategorie], 1)

    return summe

def num_cat(daten_tab):
    number = {}
    for eintrag in daten_tab:
        kategorie = eintrag["kategorie"]

        number[kategorie] = number.get(kategorie, 0) + 1
        
    return number




block = """01.03.2026;Lebensmittel;42.50
03.03.2026;Transport;19.90
05.03.2026;Lebensmittel;7.20
12.04.2026;Miete;850.00
14.04.2026;Transport;19.90
"""
for eintrag in parse_ergebnis(block):
    print(eintrag)
```
