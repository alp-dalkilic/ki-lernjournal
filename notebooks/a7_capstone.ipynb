```python
def belegliste(block):
    basis_liste = []
    for line in block.splitlines():
        if line.strip() == '':
            continue

        line = line.split(';')    
        basis_liste.append(line)

    belege = []

    for i in range(len(basis_liste)):
        datensatz = {}
        datensatz['datum'] = basis_liste[i][0]
        datensatz['kategorie'] = basis_liste[i][1]
        datensatz['betrag'] = float(basis_liste[i][2])
        belege.append(datensatz)

    return belege

block = """01.03.2026;Lebensmittel;42.50
03.03.2026;Transport;19.90
05.03.2026;Lebensmittel;7.20
12.04.2026;Miete;850.00
14.04.2026;Transport;19.90
"""
ergebnis = belegliste(block)
print(ergebnis)
```
