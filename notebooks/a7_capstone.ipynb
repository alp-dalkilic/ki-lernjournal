```python
def parser(block):
    basis_list = []
    for line in block.splitlines():
        if line.strip() == '':
            continue

        felder = line.split(';')    
        basis_list.append(felder)

    belege = []

    for datum, kategorie, betrag in basis_list:
        datensatz = {'datum': datum, 'kategorie': kategorie, 'betrag': float(betrag)}
        belege.append(datensatz)

    return belege

def summe_je_kategorie(belege):
    kategorie_summe = {}
    for beleg in belege:
        kat = beleg['kategorie']
        betrag = beleg['betrag']

        if kat in kategorie_summe:
            kategorie_summe[kat] += betrag
        else:
            kategorie_summe[kat] = betrag

    return kategorie_summe

def anzahl_je_kategorie(belege):
    kategorie_anzahl = {}
    for beleg in belege:
        kat = beleg['kategorie']

        if kat in kategorie_anzahl:
            kategorie_anzahl[kat] += 1
        else:
            kategorie_anzahl[kat] = 1

    return kategorie_anzahl

def summe_je_monat(belege):
    monat_summe = {}
    for beleg in belege:
        pre_list = beleg['datum'].split('.')
        monat = '.'.join(pre_list[1:])

        betrag = beleg['betrag']

        if monat in monat_summe:
            monat_summe[monat] += betrag
        else:
            monat_summe[monat] = betrag

    return monat_summe


block = """01.03.2026;Lebensmittel;42.50
03.03.2026;Transport;19.90
05.03.2026;Lebensmittel;7.20
12.04.2026;Miete;850.00
14.04.2026;Transport;19.90
"""
belege = parser(block)
print(f'Parser: {belege}')

summe_kat = summe_je_kategorie(belege)
print(f'Summe pro Kategorie: {summe_kat}')

anzahl_kat = anzahl_je_kategorie(belege)
print(f'Anzahl je Kategorie: {anzahl_kat}')

summe_monat = summe_je_monat(belege)
print(f'Summe pro Monat: {summe_monat}')
```
