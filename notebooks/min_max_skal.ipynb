```python
import numpy as np

X = np.array([[1.0, 100.0],
              [2.0, 200.0],
              [3.0, 300.0],
              [4.0, 400.0]])

class MinMaxSkalierer:
    def __init__(self):
        self.min_ = None
        self.max_ = None

    def fit(self, X):
        self.min_ = np.min(X, axis=0)
        self.max_ = np.max(X, axis=0)
        return self

    def transform(self, X):
        if self.min_ is None or self.max_ is None:
            raise ValueError('Zuerst fit(X) aufrufen')
        return (X - self.min_) / (self.max_ - self.min_)

    def fit_transform(self, X):
        self.fit(X)
        return self.transform(X)

s = MinMaxSkalierer()

try:
    s.transform(X)
except ValueError as e:
    print("OK, Fehler wie erwartet:")

s.fit(X)
print("min:", s.min_)        # erwartet [  1. 100.]
print("max:", s.max_)        # erwartet [  4. 400.]
print(s.transform(X))        # erwartet Spalten von 0 bis 1 in Schritten von 1/3

X_neu = np.array([[2.5, 250.0],
                  [5.0, 50.0]])
print(s.transform(X_neu))    # erwartet [[0.5 0.5] [1.333 -0.167]]

print(MinMaxSkalierer().fit_transform(X)[0])   # erwartet [0. 0.]
```
