```python
# Kata: FizzBuzzu

numbers = range(1,31)
for number in numbers:
    if number % 3 == 0 and number % 5 == 0:
        print(number, " FizzBuzz")
    elif number % 3 == 0:
        print(number, " Fizz" )
    elif number % 5 == 0:
        print(number, " Buzz")
    else:
        print(number)
```

```python
# Kata: Liste drehen

liste = [2,4,6,8,1,3,5,7,9]
liste_inverse = []

for i in range(len(liste)):
    liste_inverse.append(liste[-1-i])

print(liste_inverse)
```

```python
# Kata: Palindrom

text = "Anna"
text_lc = text.lower()
liste = list(text_lc)
liste_r = list(reversed(liste))

if liste == liste_r:
    print(f"{text} ist ein Palindrom!")
else:
    print(f"{text} ist KEIN Palindrom!")
```
