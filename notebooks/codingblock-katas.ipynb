```python
# Kata: FizzBuzzu

numbers = range(1,31)
for number in numbers:
    if number % 3 == 0 and number % 5 == 0:
        print(number, " FizzBuzz")
    elif number % 3 == 0:
        print(number, " Fizz" )
    elif number % 5 == 0:
        print(number, " Buzz")
    else:
        print(number)
```
