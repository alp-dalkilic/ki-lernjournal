```python
import numpy as np
np.set_printoptions(suppress=True)

# 4 Kredite, 3 bereits standardisierte Merkmale (wie X_train_s nach dem StandardScaler)
X = np.array([[-1.0, -0.8,  0.5],      # Kredit 0: kurz, klein, älterer Kunde
              [ 0.2,  0.1, -0.3],      # Kredit 1: durchschnittlich
              [ 1.4,  1.2, -1.1],      # Kredit 2: lang, groß, junger Kunde
              [-0.6, -0.5,  0.9]])     # Kredit 3: kurz, klein, älterer Kunde
spalten = ['laufzeit_s', 'betrag_s', 'alter_s']
y = np.array([0, 0, 1, 0])             # 1 = Ausfall

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

print(X.shape, y.shape)                # Kontrollwert: (4, 3) (4,)
```

```python
w = np.array([0.9, 0.6, -0.5])         # (3,)  — wie dein self.w
b = -0.4                               # wie dein self.b
z = X @ w + b                          # (4, 3) @ (3,) -> (4,)
a = sigmoid(z)
print('z =', z.round(3))
print('a =', a.round(3))
```

```python
w2 = np.array([-0.3, 1.1, 0.4]); b2 = 0.2
a2 = sigmoid(X @ w2 + b2)

#print(np.concatenate([a, a2]).shape)        # untereinander: (8,)  <- das war deine Antwort (falsch)
print(np.column_stack([a, a2]).shape)       # nebeneinander: (4, 2) <- das ist eine Schicht (richtig)
```

```python
W1 = np.array([[ 0.9, -0.3],     # Zeile = laufzeit_s      Spalte 0 = Neuron A (dein w)
               [ 0.6,  1.1],     # Zeile = betrag_s        Spalte 1 = Neuron B (w2)
               [-0.5,  0.4]])    # Zeile = alter_s
b1 = np.array([-0.4, 0.2])       # ein Bias je Neuron

Z1 = X @ W1 + b1                 # (4, 3) @ (3, 2) -> (4, 2); b1 wird auf jede Zeile addiert
A1 = sigmoid(Z1)
print('W1', W1.shape, ' b1', b1.shape, ' A1', A1.shape, ' Z1', Z1.shape)
print(A1.round(3))
print(np.allclose(A1[:, 0], a), np.allclose(A1[:, 1], a2))   # Spalten = die Einzel-Neuronen
```

```python
W2 = np.array([[ 1.5],           # Zeile = Eingang Neuron A (Spalte 0 von A1)
               [-0.8]])          # Zeile = Eingang Neuron B (Spalte 1 von A1)
b2 = np.array([-0.3])
Z2 = A1 @ W2 + b2                # (4, 2) @ (2, 1) -> (4, 1)
A2 = sigmoid(Z2)                 # Ausfallwahrscheinlichkeit je Kredit
print(A2.round(3))
print('Schwelle 0,5:', (A2[:, 0] >= 0.5).astype(int), ' wahr:', y)
```

```python
def dense(A_in, W, b, g):
    """Eine Schicht: alle Neuronen auf einmal. A_in (n, e), W (e, k), b (k,) -> (n, k)."""
    return g(A_in @ W + b)

def forward(X):
    A1 = dense(X,  W1, b1, sigmoid)   # Schicht 1: 3 Eingänge -> 2 Neuronen
    A2 = dense(A1, W2, b2, sigmoid)   # Schicht 2: 2 Eingänge -> 1 Neuron
    return A2

print(np.allclose(forward(X), A2))    # True
```

```python
def relu(z):
    return np.maximum(0, z)      # elementweise! np.max wäre EIN Maximum der ganzen Matrix

print(Z1.round(2))
print(relu(Z1).round(2))
```
