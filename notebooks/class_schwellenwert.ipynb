```python
import numpy as np

X = np.array([[1., 10.],
              [3., 20.],
              [5., 60.]])

class Schwellenwert:
    def __init__(self):
        self.schwelle = None

    def fit(self, X):
        self.schwelle = np.mean(X, axis=0)
        return self

    def transform(self, X):
        return (X > self.schwelle).astype(int)

    def fit_transform(self, X):
        self.fit(X)
        return self.transform(X)

s = Schwellenwert()
s.fit(X)
print(s.schwelle)        # erwartet: [ 3. 30.]
print(s.transform(X))    # erwartet: [[0 0]
                         #            [0 0]
                         #            [1 1]]

X_neu = np.array([[6., 45.],
                  [4., 35.]])
print(s.transform(X_neu))   # erwartet: [[1 0]
                            #            [0 1]]

print(Schwellenwert().fit_transform(X))   # erwartet wie transform(X) oben

```
